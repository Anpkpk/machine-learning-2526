```python
import pandas as pd
import numpy as np

data = {
    'Hoc sinh': ['HS1', 'HS2', 'HS3', 'HS4', 'HS5'],
    'Toan': [80, 70, 90, 60, 75],
    'Ly': [85, 75, 95, 65, 70],
    'Hoa': [90, 80, 85, 70, 75]
}
df = pd.DataFrame(data)
df.head()
```

```python
mean_scores = df [['Toan', 'Ly', 'Hoa']].mean()
std_scores = df [['Toan', 'Ly', 'Hoa']].std(ddof=0)
```

```python
standardized_df = df.copy()
standardized_df[['Toan', 'Ly', 'Hoa']] = (df[['Toan', 'Ly', 'Hoa']] - mean_scores) / std_scores

standardized_df.head()
```

```python
cov_matrix = standardized_df[['Toan', 'Ly', 'Hoa']].cov(ddof=0)
print("Covariance matrix:\n", cov_matrix)
```

```python
eig_val, eig_vec = np.linalg.eig(cov_matrix)
print("Eigen values:\n", eig_val)
print("Eigen vectors:\n", eig_vec)
```

```python
selected_eig_vec = eig_vec[:, [0, 2]]
reduced_data = standardized_df[['Toan', 'Ly', 'Hoa']].dot(selected_eig_vec)
reduced_data.columns = ['PC1', 'PC3']
print("Reduced data:\n", reduced_data)
```

```python
new_data = reduced_data.values
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=2, random_state=0).fit(new_data)
print("Cluster labels :", kmeans.labels_) # Result : [0 1 0 1 1]
print("Cluster centers :", kmeans.cluster_centers_)
```
